# spotify_trends_analysis

```python
import pandas as pd
import numpy as np
```

# =====================================================================
# 1. GENERATE A REALISTIC DATASET OF 10,000+ TRACKS (Simulating Raw Data)
# =====================================================================

```python
np.random.seed(42)
num_tracks = 10500
```

```python
data = {
    'track_id': [f"tr_{i}" for i in range(1, num_tracks + 1)],
    'track_name': [f"Song {i}" for i in range(1, num_tracks + 1)],
    'genre': np.random.choice(['Pop', 'Hip-Hop', 'Rock', 'Indie', 'Electronic', 'R&B'], num_tracks),
    'danceability': np.random.uniform(0.1, 1.0, num_tracks),
    'energy': np.random.uniform(0.1, 1.0, num_tracks),
    'loudness': np.random.uniform(-20, -2, num_tracks),
    'acousticness': np.random.uniform(0.0, 1.0, num_tracks),
    # Simulating a user engagement/popularity metric (0 to 100)
    'popularity_score': np.random.randint(10, 101, num_tracks),
    # Simulating a feature retention/completion metric (percentage of song listened to)
    'completion_rate_pct': np.random.uniform(15.0, 98.0, num_tracks)
}

df = pd.DataFrame(data)
```

```python
# Injecting artificial missing values (NaNs) to demonstrate data preprocessing skills
```

```python
df.loc[df.sample(frac=0.02).index, 'danceability'] = np.nan
df.loc[df.sample(frac=0.015).index, 'popularity_score'] = np.nan

print(f"📊 Initialized raw dataset with {df.shape[0]} tracks and {df.shape[1]} features.")
```

# =====================================================================
# 2. DATA PREPROCESSING & HANDLING MISSING VALUES (Pandas)
# =====================================================================

```python
# Check for missing values before cleaning
print("\nMissing values per column before preprocessing:\n", df.isnull().sum())
```

```python
# Handle missing values: Fill continuous features with the median, drop rows missing critical targets
df['danceability'] = df['danceability'].fillna(df['danceability'].median())
df = df.dropna(subset=['popularity_score'])
```

```python
print(f"\n✅ Preprocessing Complete! Cleaned dataset contains {df.shape[0]} tracks.")
```

# =====================================================================
# 3. EXPLORATORY DATA ANALYSIS (EDA) USING NUMPY & PANDAS
# =====================================================================

```python
print("\n📈 Comprehensive Statistical Summary Metrics:")
```

```python
# Calculating the exact mean, median, and variance metrics claimed on your resume
metrics_summary = pd.DataFrame({
    'Mean': df[['danceability', 'energy', 'popularity_score', 'completion_rate_pct']].mean(),
    'Median': df[['danceability', 'energy', 'popularity_score', 'completion_rate_pct']].median(),
    'Variance': df[['danceability', 'energy', 'popularity_score', 'completion_rate_pct']].var()
})
print(metrics_summary)
```

# =====================================================================
# 4. ANALYZING CORRELATION PATTERNS DRIVING POPULARITY & RETENTION
# =====================================================================

```python
print("\n🔍 Correlation Matrix (Analyzing Feature Trends):")
```

```python
# Isolating correlation metrics between track attributes and user interaction metrics
correlation_matrix = df[['danceability', 'energy', 'loudness', 'acousticness', 'popularity_score', 'completion_rate_pct']].corr()
print(correlation_matrix)
```

```python
# Grouping by genre to identify product-level performance trends
print("\n🎶 Average Performance Metrics by Genre:")
genre_trends = df.groupby('genre')[['popularity_score', 'completion_rate_pct']].mean().sort_values(by='popularity_score', ascending=False)
print(genre_trends)
```
